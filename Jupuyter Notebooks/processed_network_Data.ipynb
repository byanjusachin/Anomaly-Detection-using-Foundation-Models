### Dataset obtained from https://www.kaggle.com/datasets/dnkumars/cybersecurity-intrusion-detection-dataset/data

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
df = pd.read_csv("cybersecurity_intrusion_data.csv")
```

```python
df.head()
```

```python
# Check for null values.

null_val = df.isnull().sum()
print (null_val)
```

```python
# Check for null in text data
null_text = df['encryption_used'].apply(lambda x: isinstance(x, str) and x.strip() == '').sum()
print(f'Null values in Encruption Used Column: {null_text}')
```

```python
# Session ID is a identified assigned during a browser session.
# Session ID is unique for each brower session, thus does not provide value for learning algorithms.

df = df.drop(columns=['session_id'])
```

```python
# Handling categorical data

categorical_cols = ['protocol_type', 'encryption_used', 'browser_type']

df_encoded = pd.get_dummies(df, columns=categorical_cols, dummy_na=True, dtype=int)

print(df_encoded.columns)
print(df_encoded.dtypes)
```

```python
# Normalize the numerical features.
# Since the values for this columns do not lie withing a specific range, min-max scaler is not used.
# Applied standard scaling.

from sklearn.preprocessing import StandardScaler
columns_to_scale = ['network_packet_size', 'session_duration', 'ip_reputation_score']
scaler = StandardScaler()
df_encoded[columns_to_scale] = scaler.fit_transform(df_encoded[columns_to_scale])
print(df_encoded.head())
```

```python
# Export the dataset
df_encoded.to_csv('encoded_dataset.csv', index=False) 
```
